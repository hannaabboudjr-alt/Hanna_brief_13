```python
import pandas as pd
import numpy as np

df_activite = pd.read_csv("arrets-de-travail-par-secteur-d-activite-naf-1-et-naf-5.csv", sep=";")

print("Loaded successfully! Shape:", df_activite)
```

```python
df_activite
```

```python
df_activite.describe()
```

```python
df_activite.drop(columns=["NAF1","code_NAF1","libelle_NAF1","NAF5","nb_jours_arret_travail_txt"], inplace=True)
```

```python
df_activite
```

```python
#supprimer les ligne où risuqe = tous risuqe confondus et où code_NAF5 = xxxxx
# Filtrage AVANT de supprimer la colonne "risque"
df_activite = df_activite[
    (df_activite["risque"] != "Tous risques confondus") &
    (df_activite["code_NAF5"] != "xxxxx")
]

# Suppression des lignes "agrégées" ou non pertinentes :
df_activite = df_activite[
    (df_activite["risque"] != "Tous risques confondus") &      # on exclut les lignes "total tous risques"
    (~df_activite["code_NAF5"].isin(["xxxxx", "TOUS"]))        # on exclut les codes secteur agrégés/manquants
]
```

```python
df_activite
```

```python
df_activite.shape
```

```python
print("Le pourcentage des valeures manquantes \n",(df_activite.isna().mean() * 100).round(1).sort_values(ascending=False),"\n\n") 
```

```python
print("Le pourcentage des valeures manquantes \n",(df_cause.isna().mean() * 100).round(1).sort_values(ascending=False),"\n\n") 
```

```python
df_cause.drop(columns=["NAF1","code_NAF1","libelle_NAF1","NAF5","libelle_type_vehicule","part_cause_TR","part_cause_tr_txt"], inplace=True)
```

```python
df_cause
```

```python
# Filtrage AVANT de supprimer la colonne "risque"
df_cause = df_cause[
    
    (df_cause["code_NAF5"] != "xxxxx")
]

# Suppression des lignes "agrégées" ou non pertinentes :
df_cause = df_cause[
         (~df_cause["code_NAF5"].isin(["xxxxx", "TOUS"]))] 
```

```python
df_cause
```

```python
#supprimer les lignes où libelle_cause_TR = "Toutes causes confondues"
df_cause = df_cause[df_cause["libelle_cause_TR"] != "Toutes causes confondues"]
df_cause
```

```python


# fusionner les lignes où libelle_cause_TR = "Perte de contrôle d'un véhicule" en une seule ligne par groupe
df_cause = df_cause.drop_duplicates()
df_cause
```
